## 1. 셀의 종류
- **Code 셀**: 파이썬 코드 실행 (`In [ ]:` 표시)
- **Markdown 셀**: 설명 글. 더블클릭하면 편집, `Shift+Enter`로 렌더링

## 2. 필수 단축키
| 동작 | 키 |
|---|---|
| 셀 실행 후 다음 셀로 | `Shift+Enter` |
| 셀 실행(이동 안 함) | `Ctrl+Enter` |
| 명령 모드로 전환 | `Esc` (셀 테두리가 파랑) |
| 편집 모드로 전환 | `Enter` (테두리가 초록) |
| 아래/위에 셀 추가 | 명령 모드에서 `B` / `A` |
| 셀 삭제 | 명령 모드에서 `D` `D` |
| 코드↔마크다운 변환 | 명령 모드에서 `Y` / `M` |
| **저장** | `Ctrl+S` (시험 중 수시로!) |
| 자동완성/도움말 | `Tab` / `Shift+Tab` |

```python
print("Shift+Enter 로 실행해보세요")
1 + 1   # 셀의 마지막 줄 값은 자동으로 출력됩니다
```

## 3. 가장 흔한 실수
1. **셀 실행 순서가 꼬임**: 위에서 아래로 실행해야 변수가 정의됩니다. 이상하면 메뉴 `Kernel → Restart & Run All`로 처음부터 다시 돌려 확인하세요. (최종 제출 전에 꼭 한 번!)
2. **저장 안 함**: 시험 환경에서는 브라우저를 닫으면 저장 안 된 내용이 사라집니다. `Ctrl+S`.
3. **`In [*]`**: 별표는 실행 중이라는 뜻입니다. 오래 걸리면 `Kernel → Interrupt`(■ 버튼)로 중단.
4. **초기화 금지**: 시험 화면의 '초기화'는 작성 내용을 전부 지웁니다.

## 4. 파일 읽기/쓰기 연습 (제출 csv 저장 방식)

```python
import pandas as pd
import os

df = pd.DataFrame({"id": [1, 2, 3], "pred": [0, 1, 0]})
os.makedirs("out", exist_ok=True)
df.to_csv("out/submission.csv", index=False)   # index=False 필수! (안 하면 불필요한 열이 생김)
print(pd.read_csv("out/submission.csv"))
print(os.listdir("out"))
```

## 5. 유용한 매직 명령

- `!pip install 패키지` : 패키지 설치 (설치 후 `Kernel → Restart` 필요할 수 있음)
- `!ls` : 현재 폴더 파일 목록
